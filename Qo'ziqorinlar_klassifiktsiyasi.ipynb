**Mushroom dataset ustunlari va ularning tavsifi**  

1. **class** – qoʻziqorinning turi: zaharli (`p`) yoki zaharsiz (`e`). *Bu target (maqsad) ustun hisoblanadi.*  
2. **cap-shape** – qoʻziqorin shapkasi shakli:  
   - `b`: bell (qo‘ng‘iroq)  
   - `c`: conical (konussimon)  
   - `x`: convex (yumaloq)  
   - `f`: flat (tekis)  
   - `k`: knobbed (dumaloq tugunli)  
   - `s`: sunken (botiq)  

3. **cap-surface** – qoʻziqorin shapkasi yuzasi qanday koʻrinishda:  
   - `f`: fibrous (tolali)  
   - `g`: grooved (egik chiziqli)  
   - `y`: scaly (taroqli)  
   - `s`: smooth (silliq)  

4. **cap-color** – qoʻziqorin shapkasi rangi:

   - `u`: purple (binafsha)

   - `y`: yellow (sariq)  

5. **bruises** – qoʻziqorin bosilganda dogʻ qoldiradimi?

6. **odor** – qoʻziqorin hidi:

7. **gill-attachment** – qoʻziqorin plastinkalarining shapkaga bogʻlanish turi:

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.preprocessing import OrdinalEncoder,LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python


url="https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/mushrooms.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.columns=df.columns.map(lambda x:x.strip())
df.columns=df.columns.map(lambda x:x.replace("-","_"))
```

```python
# barcha ustunlarni vizuval trzda ko'rib chiqamiz
for col in df.columns:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=col)
    plt.title(f"{col} tarqalishi")
    plt.grid()
    plt.tight_layout()
    plt.show()
```

```python
# Y ustunimiz muozanatini tekshiramiz

sns.countplot(data=df, x="class")
plt.title("Zaharli va Zararsiz qo'ziqorinlar")
plt.show()

# Klasslar soni
d = df["class"].value_counts()
print("zaxaerli va zaxarsizlarning soni:\n", d)

# Tafsilotli chiqish
if "e" in df["class"].values:
    print("Zararsiz qo'ziqorinlar soni:", len(df[df["class"] == "e"]))

if "p" in df["class"].values:
    print("Zaharli qo'ziqorinlar soni:", len(df[df["class"] == "p"]))

# tafovut 290 dona eka , bu tafovvut 8124 qatorlik modelga  jiddiy tasir qilmaydi
```

```python
df.info()
```

```python
df.describe()
```

```python
# df ni bolib olamiz
train_set,test_set=train_test_split(df,test_size=0.3,random_state=42)
```

```python
# ML ga tayyorgarlik
x=train_set.drop("class",axis=1)
y=train_set[["class"]]

ord=OrdinalEncoder()
leb=LabelEncoder()
x=ord.fit_transform(x)
y=leb.fit_transform(y)
```

```python
# start ML
#knn algoritimidan foydalanamiz
knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(x, y)

```

```python
# test_set ni ML ga tayyorlash
x_test=test_set.drop("class",axis=1)
y_test=test_set[["class"]]

ord=OrdinalEncoder()
leb=LabelEncoder()
x_test=ord.fit_transform(x_test)
y_test=leb.fit_transform(y_test)
```

```python
# Bashorat qilish
y_pred = knn.predict(x_test)
```

```python
# tekshiruv ,  kriteryalar bo'yicha
print(classification_report(y_pred,y_test))
```

```python
# matix bo'yicha tekshiruv
print(confusion_matrix(y_test,y_pred))
```

```python

report = classification_report(y_test, y_pred, output_dict=True)

df_report = pd.DataFrame(report).transpose()


df_plot = df_report[["precision", "recall", "f1-score"]].iloc[:2]

df_plot.plot(kind="bar", figsize=(8, 6))
plt.title("Classification Report Metrics")
plt.xlabel("Class")
plt.ylabel("Score")
plt.ylim(0, 1.1)
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

```python
sns.heatmap(confusion_matrix(y_test,y_pred), annot=True)
plt.show()
# KNN algoritmda natija judayaxshi chiqdi
```

```python
# keyingi algoritmlarda ham sinab ko'tamiz
```

```python
# Decision Tree algoritimi ML qismi
dt=DecisionTreeClassifier(random_state=42)
dt.fit(x,y)
```

```python
# D TREE algoritimi bashorat qismi
y_predd=dt.predict(x_test)
```

```python
report = classification_report(y_test, y_predd, output_dict=True)

df_report = pd.DataFrame(report).transpose()


df_plot = df_report[["precision", "recall", "f1-score"]].iloc[:2]

df_plot.plot(kind="bar", figsize=(8, 6))
plt.title("Classification Report Metrics")
plt.xlabel("Class")
plt.ylabel("Score")
plt.ylim(0, 1.1)
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

```python
sns.heatmap(confusion_matrix(y_test,y_predd), annot=True)
plt.show()
```

```python
print(classification_report(y_predd,y_test))
```

```python
print("aniqlik darajasi",accuracy_score(y_predd,y_test))
```

```python
# model 2 ta algoritmda ham ideal natija berdi
# bu xolatdan shubxalanib cross validation qilib ko'rmoqchiman

# 5-fold cross-validation orqali baholash
scores = cross_val_score(dt, x, y, cv=5)
print("Cross-val accuracy:", scores)
print("O'rtacha:", scores.mean())
```
